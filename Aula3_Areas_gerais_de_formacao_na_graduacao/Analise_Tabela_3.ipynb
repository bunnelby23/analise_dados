# Atividade — Análise da Tabela 3

A Tabela 3 foi dividida em três tabelas por região:

1. Homens e mulheres com curso superior;
2. Homens e mulheres com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática (CTEM);
3. Homens e mulheres com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar.

As três tabelas foram salvas em arquivos `.xlsx` e, ao final, são apresentadas as estatísticas descritivas com `describe()`.

```python
import pandas as pd
import openpyxl

path = r"Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
dados = pd.read_excel(path, header=None)
dados.head()
```

## 1. Homens e mulheres por região com curso superior

```python
regioes = dados.iloc[4:9].copy()
curso_superior = regioes.iloc[:, [0, 2, 3]].copy()
curso_superior.columns = ["Região", "Homens", "Mulheres"]
curso_superior
```

```python
curso_superior.to_excel(r"curso_superior_por_regiao.xlsx", index=False)
```

### Estatísticas descritivas

```python
curso_superior.describe()
```

## 2. Homens e mulheres por região — Ciência, Tecnologia, Engenharias e Matemática

```python
ctem = regioes.iloc[:, [0, 5, 6]].copy()
ctem.columns = ["Região", "Homens", "Mulheres"]
ctem
```

```python
ctem.to_excel(r"ctem_por_regiao.xlsx", index=False)
```

### Estatísticas descritivas

```python
ctem.describe()
```

## 3. Homens e mulheres por região — Educação, Serviços pessoais, Saúde e Bem-estar

```python
educacao_saude = regioes.iloc[:, [0, 8, 9]].copy()
educacao_saude.columns = ["Região", "Homens", "Mulheres"]
educacao_saude
```

```python
educacao_saude.to_excel(r"educacao_servicos_saude_bem_estar_por_regiao.xlsx", index=False)
```

### Estatísticas descritivas

```python
educacao_saude.describe()
```
