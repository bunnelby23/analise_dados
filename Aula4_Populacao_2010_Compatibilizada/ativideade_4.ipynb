# Atividade 4 — Análise da população

```python
import pandas as pd
from pathlib import Path

PASTA_PROJETO = Path.cwd()
ARQUIVO = PASTA_PROJETO / "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

df = pd.read_excel(ARQUIVO, header=2)

df = df[[
    "UF",
    "COD. UF",
    "COD. MUNIC",
    "NOME DO MUNICÍPIO",
    "População Município 2010\n(Sinopse)",
    "População 2010 (Alterações de Limites até 2022)1",
    "População Censo 2022"
]].copy()

df = df.dropna(subset=["UF", "NOME DO MUNICÍPIO"])
df["População 2010 (Alterações de Limites até 2022)1"] = pd.to_numeric(
    df["População 2010 (Alterações de Limites até 2022)1"], errors="coerce"
)
df["População Censo 2022"] = pd.to_numeric(
    df["População Censo 2022"], errors="coerce"
)

df.head()
```

## 1. População agregada por estado


```python
populacao_estado = (
    df.groupby("UF")[[
        "População 2010 (Alterações de Limites até 2022)1",
        "População Censo 2022"
    ]]
    .sum()
    .reset_index()
)

populacao_estado["Crescimento 2022 - 2010"] = (
    populacao_estado["População Censo 2022"]
    - populacao_estado["População 2010 (Alterações de Limites até 2022)1"]
)

populacao_estado = populacao_estado.sort_values(
    "Crescimento 2022 - 2010", ascending=False
).reset_index(drop=True)

populacao_estado
```

```python
arquivo_estado = PASTA_PROJETO / "populacao_por_estado_crescimento.csv"
populacao_estado.to_csv(arquivo_estado, sep=";", index=False, encoding="utf-8-sig")
print(f"Arquivo salvo em: {arquivo_estado}")
```

## 2. População por município

```python
populacao_municipio = df[[
    "UF",
    "COD. UF",
    "COD. MUNIC",
    "NOME DO MUNICÍPIO",
    "População 2010 (Alterações de Limites até 2022)1",
    "População Censo 2022"
]].copy()

populacao_municipio["Crescimento 2022 - 2010"] = (
    populacao_municipio["População Censo 2022"]
    - populacao_municipio["População 2010 (Alterações de Limites até 2022)1"]
)

populacao_municipio = populacao_municipio.sort_values(
    "Crescimento 2022 - 2010", ascending=False
).reset_index(drop=True)

populacao_municipio
```

```python
arquivo_municipio = PASTA_PROJETO / "populacao_por_municipio_crescimento.csv"
populacao_municipio.to_csv(arquivo_municipio, sep=";", index=False, encoding="utf-8-sig")
print(f"Arquivo salvo em: {arquivo_municipio}")
```
